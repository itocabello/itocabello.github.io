```python
%%writefile mapper_organizations.py
#!/usr/bin/env python3

import sys
import csv

# Lectura del CSV desde la entrada estándar
reader = csv.DictReader(sys.stdin)

for row in reader:

    city = row.get("CITY", "").strip()
    state = row.get("STATE", "").strip()

    # Comprobar que existen los datos necesarios
    if not city or not state:
        continue

    # Normalizar los valores
    city = city.strip()
    state = state.strip()

    # Crear clave compuesta: estado | ciudad
    key = f"{state}|{city}"

    # Emitir un registro por organización
    print(f"{key}\t1")



```
