```python
%%writefile /media/notebooks/EjercicioFinal/join_clinico_avro.hql
CREATE TABLE join_clinico_avro
STORED AS AVRO
AS
SELECT
    e.Id AS encounter_id,

    -- Datos del paciente
    p.Id AS patient_id,
    p.FIRST,
    p.LAST,
    p.BIRTHDATE,
    p.DEATHDATE,
    p.GENDER,

    -- Datos del encuentro
    e.`START`,
    e.`STOP`,
    e.ENCOUNTERCLASS,
    e.CODE,
    e.DESCRIPTION,
    e.BASE_ENCOUNTER_COST,
    e.TOTAL_CLAIM_COST,
    e.PAYER_COVERAGE,

    -- Datos de la organización
    o.Id AS organization_id,
    o.NAME AS organization_name,
    o.ADDRESS,
    o.CITY,
    o.STATE,
    o.ZIP

FROM encounters_avro e

INNER JOIN patients_avro p
    ON e.PATIENT = p.Id

INNER JOIN organizations_avro o
    ON e.ORGANIZATION = o.Id;
```

```python
%%writefile /media/notebooks/EjercicioFinal/join_clinico_parquet.hql
CREATE TABLE join_clinico_parquet
STORED AS PARQUET
AS
SELECT
    e.Id AS encounter_id,

    p.Id AS patient_id,
    p.FIRST,
    p.LAST,
    p.BIRTHDATE,
    p.DEATHDATE,
    p.GENDER,

    e.START_DATE,
    e.STOP_DATE,
    e.ENCOUNTERCLASS,
    e.CODE,
    e.DESCRIPTION,
    e.BASE_ENCOUNTER_COST,
    e.TOTAL_CLAIM_COST,
    e.PAYER_COVERAGE,

    o.Id AS organization_id,
    o.NAME AS organization_name,
    o.ADDRESS,
    o.CITY,
    o.STATE,
    o.ZIP

FROM encounters_parquet e
INNER JOIN patients_parquet p
    ON e.PATIENT = p.Id
INNER JOIN organizations_parquet o
    ON e.ORGANIZATION = o.Id;
```

```python
%%writefile /media/notebooks/EjercicioFinal/MonitorProceso.py
import subprocess
import time
import psutil


class MonitorProceso:

    def __init__(self, script):
        self.script = script
        self.tiempo = 0
        self.memoria_maxima = 0

    def ejecutar(self):

        inicio = time.perf_counter()

        proceso = subprocess.Popen(
            ["bash", self.script],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.PIPE
        )

        proceso_psutil = psutil.Process(proceso.pid)

        while proceso.poll() is None:

            try:
                memoria_actual = (
                    proceso_psutil.memory_info().rss
                    / (1024 * 1024)
                )

                self.memoria_maxima = max(
                    self.memoria_maxima,
                    memoria_actual
                )

            except psutil.NoSuchProcess:
                break

            time.sleep(0.5)

        fin = time.perf_counter()

        self.tiempo = fin - inicio

        return {
            "script": self.script,
            "tiempo": self.tiempo,
            "memoria": self.memoria_maxima,
            "codigo_salida": proceso.returncode
        }
```

```python
%%writefile /media/notebooks/EjercicioFinal/medir_procesos.py
from MonitorProceso import MonitorProceso


procesos = [
    ("MapReduce CSV", "mapreduce_csv.sh"),
    ("MapReduce Parquet", "mapreduce_parquet.sh"),
    ("MapReduce Avro", "mapreduce_avro.sh"),
    ("Hive CSV", "hive_csv.sh"),
    ("Hive Parquet", "hive_parquet.sh"),
    ("Hive Avro", "hive_avro.sh")
]


resultados = []


for nombre, script in procesos:

    print(f"\nEjecutando: {nombre}")

    monitor = MonitorProceso(script)

    resultado = monitor.ejecutar()

    resultados.append({
        "proceso": nombre,
        "tiempo": resultado["tiempo"],
        "memoria": resultado["memoria"],
        "codigo_salida": resultado["codigo_salida"]
    })

    print(
        f"Tiempo de ejecución: "
        f"{resultado['tiempo']:.2f} segundos"
    )

    print(
        f"Memoria máxima: "
        f"{resultado['memoria']:.2f} MB"
    )

    print(
        f"Código de salida: "
        f"{resultado['codigo_salida']}"
    )
```
