```python
# Visualizar los datos en avro
from fastavro import reader

with open("/media/notebooks/dataset/TABLA_ACCIDENTES_24.avro", "rb") as f:
    avro_reader = reader(f)
    records = list(avro_reader)

records[:5] 
```

```python
# Visualizar los datos en parquet
import pandas as pd

# Leer fichero CSV
df = pd.read_csv(
    "/media/notebooks/dataset/TABLA_ACCIDENTES_24.csv",
    sep=";",
    encoding="latin1",
    low_memory=False
)

# Sustituir valores nulos
df = df.where(pd.notnull(df), None)


# Exportar a formato Parquet
df.to_parquet(
    "/media/notebooks/dataset/TABLA_ACCIDENTES_24.parquet",
    engine="pyarrow",
    compression="snappy",
    index=False
)

print("Conversión a PARQUET finalizada correctamente.")


df_parquet = pd.read_parquet(
    "TABLA_ACCIDENTES_24.parquet",
    engine="pyarrow"
)

print(df_parquet.head())
```

```python
from fastparquet import ParquetFile

pf = ParquetFile("/media/notebooks/dataset/TABLA_ACCIDENTES_24.parquet")

print(pf.schema)
```
