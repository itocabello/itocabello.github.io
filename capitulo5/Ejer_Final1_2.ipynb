```python
%%writefile convertir_txt_csv.py
#!/usr/bin/env python3

import csv
import sys
import os


def convertir_txt_csv(archivo_txt, archivo_csv):
    """
    Convierte un fichero TXT delimitado en un fichero CSV.

    Parámetros:
        archivo_txt: ruta del fichero TXT de entrada.
        archivo_csv: ruta del fichero CSV de salida.
    """

    try:
        with open(archivo_txt, "r", encoding="utf-8-sig") as entrada:
            contenido = entrada.readlines()

        with open(archivo_csv, "w", encoding="utf-8", newline="") as salida:
            escritor = csv.writer(salida)

            for linea in contenido:
                linea = linea.strip()

                if not linea:
                    continue

                # Detectar automáticamente el separador
                if ";" in linea:
                    campos = linea.split(";")
                elif "\t" in linea:
                    campos = linea.split("\t")
                elif "|" in linea:
                    campos = linea.split("|")
                else:
                    campos = linea.split(",")

                escritor.writerow([campo.strip() for campo in campos])

        print("Conversión realizada correctamente.")
        print(f"Entrada : {archivo_txt}")
        print(f"Salida  : {archivo_csv}")

    except FileNotFoundError:
        print(f"Error: no se encuentra el fichero {archivo_txt}")
        sys.exit(1)

    except Exception as e:
        print(f"Error durante la conversión: {e}")
        sys.exit(1)


if __name__ == "__main__":

    if len(sys.argv) != 3:
        print("Uso:")
        print("python3 convertir_txt_csv.py archivo.txt archivo.csv")
        sys.exit(1)

    archivo_txt = sys.argv[1]
    archivo_csv = sys.argv[2]
```
