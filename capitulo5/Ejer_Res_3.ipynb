```python
import pandas as pd

# Leer fichero CSV
df = pd.read_csv(
    "/media/notebooks/dataset/TABLA_ACCIDENTES_24.csv",
    sep=";",
    encoding="latin1",
    low_memory=False
)

# Sustituir valores nulos
df = df.where(pd.notnull(df), None)


# Exportar a formato Parquet
df.to_parquet(
    "TABLA_ACCIDENTES_24.parquet",
    engine="pyarrow",
    compression="snappy",
    index=False
)

print("Conversión a PARQUET finalizada correctamente.")


df_parquet = pd.read_parquet(
    "TABLA_ACCIDENTES_24.parquet",
    engine="pyarrow"
)

print(df_parquet.head())
```
