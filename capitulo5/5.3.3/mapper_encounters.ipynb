````python
%%writefile mapper_encounters.py
#!/usr/bin/env python3

import sys
import csv

# Lectura del CSV desde la entrada estándar
reader = csv.DictReader(sys.stdin)

for row in reader:

    start = row.get("START", "").strip()
    encounter_class = row.get("ENCOUNTERCLASS", "").strip()
    total_claim_cost = row.get("TOTAL_CLAIM_COST", "").strip()

    # Comprobar que existen los datos necesarios
    if not start or not encounter_class or not total_claim_cost:
        continue

    try:
        # Obtener el año de la fecha del encuentro
        year = start[:4]

        # Convertir el coste a número
        cost = float(total_claim_cost)

        # Clave compuesta: año | tipo de encuentro
        key = f"{year}|{encounter_class}"

        # Emitir coste y contador
        print(f"{key}\t{cost}\t1")

    except ValueError:
        continue
```


````
