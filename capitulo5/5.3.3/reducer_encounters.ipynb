```python
%%writefile reducer_pacientes.py
#!/usr/bin/env python3

import sys

current_key = None
total_cost = 0.0
count = 0

for line in sys.stdin:

    line = line.strip()

    if not line:
        continue

    try:
        key, cost, number = line.split("\t")

        cost = float(cost)
        number = int(number)

    except ValueError:
        continue

    # Primera clave
    if current_key is None:

        current_key = key
        total_cost = cost
        count = number

    # Misma clave
    elif key == current_key:

        total_cost += cost
        count += number

    # Nueva clave
    else:

        average = total_cost / count if count > 0 else 0

        year, encounter_class = current_key.split("|", 1)

        print(
            f"{year}\t{encounter_class}\t"
            f"{count}\t"
            f"{total_cost:.2f}\t"
            f"{average:.2f}"
        )

        current_key = key
        total_cost = cost
        count = number

# Procesar la última clave
if current_key is not None:

    average = total_cost / count if count > 0 else 0

    year, encounter_class = current_key.split("|", 1)

    print(
        f"{year}\t{encounter_class}\t"
        f"{count}\t"
        f"{total_cost:.2f}\t"
        f"{average:.2f}"
    )



```
