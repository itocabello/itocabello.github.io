```python
%%writefile reducer_pacientes.py
#!/usr/bin/env python3

import sys

current_key = None
total = 0

for line in sys.stdin:

    line = line.strip()

    if not line:
        continue

    try:
        key, value = line.split("\t", 1)
        value = int(value)
    except ValueError:
        continue

    # Primera clave
    if current_key is None:

        current_key = key
        total = value

    # Misma clave
    elif key == current_key:

        total += value

    # Nueva clave
    else:

        state, city = current_key.split("|", 1)

        print(f"{state}\t{city}\t{total}")

        current_key = key
        total = value

# Procesar la última clave
if current_key is not None:

    state, city = current_key.split("|", 1)

    print(f"{state}\t{city}\t{total}")




```
